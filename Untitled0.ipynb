```python
from abc import ABC, abstractmethod
from dataclasses import dataclass, field
from typing import Tuple, List, Optional
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

@dataclass(frozen=True)
class PhysicalConstants:
    c0: float = 299_792_458.0
    eps0: float = 8.854_187_812_8e-12
    mu0: float = 1.256_637_062_12e-6

    @property
    def eta0(self) -> float:
        return np.sqrt(self.mu0 / self.eps0)

@dataclass
class GridConfig1D:
    nx: int = 250
    dx: float = 1e-3
    courant_number: float = 1.0
    time_steps: int = 450
    constants: PhysicalConstants = field(default_factory=PhysicalConstants)

    @property
    def dt(self) -> float:
        return self.courant_number * self.dx / self.constants.c0
```

```python
class IExcitationSource(ABC):
    @abstractmethod
    def evaluate(self, time_step: int, dt: float) -> float:
        pass

class GaussianPulseSource(IExcitationSource):
    def __init__(self, position: int, t0_factor: float = 40.0, spread_factor: float = 12.0) -> None:
        self.position = position
        self.t0_factor = t0_factor
        self.spread_factor = spread_factor

    def evaluate(self, time_step: int, dt: float) -> float:
        t0 = self.t0_factor * dt
        spread = self.spread_factor * dt
        t = time_step * dt
        return float(np.exp(-((t - t0) / spread) ** 2))

class IBoundaryCondition(ABC):
    @abstractmethod
    def apply_ez_boundary(self, ez: np.ndarray, ez_history_prev: np.ndarray) -> None:
        pass

class MurABC1D(IBoundaryCondition):
    def __init__(self, grid_config: GridConfig1D) -> None:
        c0 = grid_config.constants.c0
        dt = grid_config.dt
        dx = grid_config.dx
        self.coeff = (c0 * dt - dx) / (c0 * dt + dx)

    def apply_ez_boundary(self, ez: np.ndarray, ez_history_prev: np.ndarray) -> None:
        ez[0] = ez_history_prev[1] + self.coeff * (ez[1] - ez_history_prev[0])
        ez[-1] = ez_history_prev[-2] + self.coeff * (ez[-2] - ez_history_prev[-1])
```

```python
class FDTD1DEngine:
    def __init__(
        self,
        grid_config: GridConfig1D,
        sources: Optional[List[IExcitationSource]] = None,
        boundary_condition: Optional[IBoundaryCondition] = None,
        eps_r_profile: Optional[np.ndarray] = None
    ) -> None:
        self.cfg = grid_config
        self.sources = sources or []
        self.boundary_condition = boundary_condition

        self.eps_r = eps_r_profile if eps_r_profile is not None else np.ones(self.cfg.nx, dtype=np.float64)

        self.ez = np.zeros(self.cfg.nx, dtype=np.float64)
        self.hy = np.zeros(self.cfg.nx, dtype=np.float64)

        self._ch = self.cfg.dt / (self.cfg.constants.mu0 * self.cfg.dx)
        self._ce = self.cfg.dt / (self.cfg.constants.eps0 * self.eps_r * self.cfg.dx)

        self.ez_history = np.zeros((self.cfg.time_steps, self.cfg.nx), dtype=np.float64)
        self.hy_history = np.zeros((self.cfg.time_steps, self.cfg.nx), dtype=np.float64)

    def step(self, current_step: int) -> None:
        ez_old_left = self.ez[0]
        ez_old_right = self.ez[-1]

        self.hy[:-1] += self._ch * (self.ez[1:] - self.ez[:-1])
        self.ez[1:] += self._ce[1:] * (self.hy[1:] - self.hy[:-1])

        if self.boundary_condition:
            prev_boundary = np.array([ez_old_left, ez_old_right])
            self.boundary_condition.apply_ez_boundary(self.ez, prev_boundary)

        for source in self.sources:
            if isinstance(source, GaussianPulseSource):
                self.ez[source.position] += source.evaluate(current_step, self.cfg.dt)

        self.ez_history[current_step, :] = self.ez.copy()
        self.hy_history[current_step, :] = self.hy.copy()

    def run(self) -> None:
        for n in range(self.cfg.time_steps):
            self.step(n)


class FDTDVisualizer:
    @staticmethod
    def render_html5_animation(
        engine: FDTD1DEngine,
        fps: int = 30,
        dpi: int = 100,
        embed_limit_mb: float = 100.0
    ) -> HTML:
        plt.rcParams['animation.embed_limit'] = embed_limit_mb

        cfg = engine.cfg
        fig, ax = plt.subplots(figsize=(10, 4.5), dpi=dpi)

        if not np.all(engine.eps_r == 1.0):
            ax.fill_between(
                np.arange(cfg.nx),
                -1.5, 1.5,
                where=(engine.eps_r > 1.0),
                color='gray', alpha=0.2, label=r'Dielectric ($\epsilon_r > 1$)'
            )

        line_ez, = ax.plot([], [], color='#D62728', lw=2.0, label=r'$E_z$ (V/m)')
        line_hy, = ax.plot([], [], color='#1F77B4', lw=1.5, linestyle='--', label=r'$H_y \times Z_0$')

        ax.set_ylim(-1.2, 1.2)
        ax.set_xlim(0, cfg.nx)
        ax.set_xlabel("Spatial Grid Index ($x$)", fontsize=11)
        ax.set_ylabel("Field Amplitude", fontsize=11)
        ax.grid(True, linestyle='--', alpha=0.5)
        ax.legend(loc='upper right', framealpha=0.9)

        def animate(frame: int) -> Tuple:
            line_ez.set_data(np.arange(cfg.nx), engine.ez_history[frame])
            line_hy.set_data(np.arange(cfg.nx), engine.hy_history[frame] * cfg.constants.eta0)
            ax.set_title(f"1D FDTD Simulation — Step: {frame} / {cfg.time_steps}", fontsize=12)
            return line_ez, line_hy

        anim = FuncAnimation(
            fig,
            animate,
            frames=cfg.time_steps,
            interval=1000 // fps,
            blit=True
        )
        plt.close(fig)

        return HTML(anim.to_jshtml())

    @staticmethod
    def render_plotly_interactive(engine: FDTD1DEngine, step_stride: int = 5) -> go.Figure:
        cfg = engine.cfg
        x = np.arange(cfg.nx)

        fig = go.Figure()

        if not np.all(engine.eps_r == 1.0):
            dielectric_mask = engine.eps_r > 1.0
            fig.add_trace(go.Scatter(
                x=x[dielectric_mask],
                y=[1.5] * np.sum(dielectric_mask),
                mode='lines',
                fill='tozeroy',
                fillcolor='rgba(128, 128, 128, 0.2)',
                line=dict(color='rgba(255,255,255,0)'),
                name='Dielectric Region',
                hoverinfo='skip'
            ))

        fig.add_trace(go.Scatter(
            x=x,
            y=engine.ez_history[0],
            mode='lines',
            line=dict(color='#D62728', width=2),
            name='Ez (V/m)'
        ))

        fig.add_trace(go.Scatter(
            x=x,
            y=engine.hy_history[0] * cfg.constants.eta0,
            mode='lines',
            line=dict(color='#1F77B4', width=1.5, dash='dash'),
            name='Hy * Z0'
        ))

        frames = []
        slider_steps = []

        for k in range(0, cfg.time_steps, step_stride):
            frame_ez = engine.ez_history[k]
            frame_hy = engine.hy_history[k] * cfg.constants.eta0

            frame_data = [go.Scatter(x=x, y=frame_ez), go.Scatter(x=x, y=frame_hy)]
            if not np.all(engine.eps_r == 1.0):
                frame_data.insert(0, go.Scatter())

            frames.append(go.Frame(
                data=frame_data,
                name=str(k)
            ))

            slider_step = dict(
                method="animate",
                args=[[str(k)], dict(mode="immediate", frame=dict(duration=30, redraw=True), transition=dict(duration=0))],
                label=f"Step {k}"
            )
            slider_steps.append(slider_step)

        fig.frames = frames

        fig.update_layout(
            title="Interactive 1D FDTD Simulation (Plotly)",
            xaxis=dict(title="Spatial Grid Index (x)", range=[0, cfg.nx]),
            yaxis=dict(title="Field Amplitude", range=[-1.2, 1.2]),
            updatemenus=[dict(
                type="buttons",
                showactive=False,
                buttons=[
                    dict(label="Play", method="animate", args=[None, dict(frame=dict(duration=30, redraw=True), fromcurrent=True)]),
                    dict(label="Pause", method="animate", args=[[None], dict(mode="immediate", frame=dict(duration=0, redraw=False))])
                ],
                x=0.1, y=1.15
            )],
            sliders=[dict(
                active=0,
                yanchor="top",
                xanchor="left",
                currentvalue=dict(font=dict(size=14), prefix="Time Step: ", visible=True, xanchor="right"),
                pad=dict(b=10, t=50),
                len=0.9,
                x=0.1,
                y=0,
                steps=slider_steps
            )]
        )

        return fig
```

```python
grid_cfg = GridConfig1D(nx=300, time_steps=500)

eps_r_profile = np.ones(grid_cfg.nx, dtype=np.float64)
eps_r_profile[180:] = 4.0

pulse_source = GaussianPulseSource(position=80)
mur_abc = MurABC1D(grid_config=grid_cfg)

sim_engine = FDTD1DEngine(
    grid_config=grid_cfg,
    sources=[pulse_source],
    boundary_condition=mur_abc,
    eps_r_profile=eps_r_profile
)
sim_engine.run()

display(FDTDVisualizer.render_html5_animation(sim_engine, fps=60, embed_limit_mb=100.0))
#fig = FDTDVisualizer.render_plotly_interactive(sim_engine, step_stride=3)
#fig.show()
```

```python
from abc import ABC, abstractmethod
from dataclasses import dataclass, field
from typing import List, Optional
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display


# =============================================================================
# 1. YAPILANDIRMA VE FİZİKSEL SABİTLER
# =============================================================================
@dataclass(frozen=True)
class PhysicalConstants:
    c0: float = 299_792_458.0
    eps0: float = 8.854_187_812_8e-12
    mu0: float = 1.256_637_062_12e-6

    @property
    def eta0(self) -> float:
        return np.sqrt(self.mu0 / self.eps0)


@dataclass
class GridConfig2D:
    nx: int = 150
    ny: int = 150
    dx: float = 1e-3
    dy: float = 1e-3
    courant_number: float = 0.5  # 2D Kararlılık Sınırı (CFL): S <= 1/sqrt(2) ≈ 0.707
    time_steps: int = 300
    constants: PhysicalConstants = field(default_factory=PhysicalConstants)

    @property
    def dt(self) -> float:
        return self.courant_number / (self.constants.c0 * np.sqrt((1.0 / self.dx)**2 + (1.0 / self.dy)**2))


# =============================================================================
# 2. STRATEJİ ARABİRİMLERİ VE SOMUT SINIFLAR (STRATEGY PATTERN)
# =============================================================================
class IExcitationSource2D(ABC):
    @abstractmethod
    def evaluate(self, time_step: int, dt: float) -> float:
        pass


class PointGaussianPulse2D(IExcitationSource2D):
    def __init__(self, pos_x: int, pos_y: int, t0_factor: float = 40.0, spread_factor: float = 12.0) -> None:
        self.pos_x = pos_x
        self.pos_y = pos_y
        self.t0_factor = t0_factor
        self.spread_factor = spread_factor

    def evaluate(self, time_step: int, dt: float) -> float:
        t0 = self.t0_factor * dt
        spread = self.spread_factor * dt
        t = time_step * dt
        return float(np.exp(-((t - t0) / spread) ** 2))


class IBoundaryCondition2D(ABC):
    @abstractmethod
    def apply_boundary(self, ez: np.ndarray, ez_prev_boundaries: dict) -> None:
        pass


class MurABC2D(IBoundaryCondition2D):
    def __init__(self, grid_config: GridConfig2D) -> None:
        c0 = grid_config.constants.c0
        dt = grid_config.dt
        dx = grid_config.dx
        self.coeff_x = (c0 * dt - dx) / (c0 * dt + dx)
        self.coeff_y = (c0 * dt - grid_config.dy) / (c0 * dt + grid_config.dy)

    def apply_boundary(self, ez: np.ndarray, ez_prev_boundaries: dict) -> None:
        ez[0, :] = ez_prev_boundaries['left'] + self.coeff_x * (ez[1, :] - ez[0, :])
        ez[-1, :] = ez_prev_boundaries['right'] + self.coeff_x * (ez[-2, :] - ez[-1, :])
        ez[:, 0] = ez_prev_boundaries['bottom'] + self.coeff_y * (ez[:, 1] - ez[:, 0])
        ez[:, -1] = ez_prev_boundaries['top'] + self.coeff_y * (ez[:, -2] - ez[:, -1])


# =============================================================================
# 3. 2D HESAP MOTORU (TMz MODU)
# =============================================================================
class FDTD2DEngineTMz:
    def __init__(
        self,
        grid_config: GridConfig2D,
        sources: Optional[List[IExcitationSource2D]] = None,
        boundary_condition: Optional[IBoundaryCondition2D] = None,
        eps_r_profile: Optional[np.ndarray] = None
    ) -> None:
        self.cfg = grid_config
        self.sources = sources or []
        self.boundary_condition = boundary_condition

        self.eps_r = eps_r_profile if eps_r_profile is not None else np.ones((self.cfg.nx, self.cfg.ny), dtype=np.float64)

        self.ez = np.zeros((self.cfg.nx, self.cfg.ny), dtype=np.float64)
        self.hx = np.zeros((self.cfg.nx, self.cfg.ny), dtype=np.float64)
        self.hy = np.zeros((self.cfg.nx, self.cfg.ny), dtype=np.float64)

        self._chx = self.cfg.dt / (self.cfg.constants.mu0 * self.cfg.dy)
        self._chy = self.cfg.dt / (self.cfg.constants.mu0 * self.cfg.dx)
        self._cez = self.cfg.dt / (self.cfg.constants.eps0 * self.eps_r)

        self.ez_history = np.zeros((self.cfg.time_steps, self.cfg.nx, self.cfg.ny), dtype=np.float64)

    def step(self, current_step: int) -> None:
        ez_prev_boundaries = {
            'left': self.ez[1, :].copy(),
            'right': self.ez[-2, :].copy(),
            'bottom': self.ez[:, 1].copy(),
            'top': self.ez[:, -2].copy()
        }

        # 1. Hx ve Hy Güncellemesi (Yee Grid 2D)
        self.hx[:, :-1] -= self._chx * (self.ez[:, 1:] - self.ez[:, :-1])
        self.hy[:-1, :] += self._chy * (self.ez[1:, :] - self.ez[:-1, :])

        # 2. Ez Güncellemesi (Kıvrılma / Curl hesabı)
        self.ez[1:, 1:] += self._cez[1:, 1:] * (
            (self.hy[1:, 1:] - self.hy[:-1, 1:]) / self.cfg.dx -
            (self.hx[1:, 1:] - self.hx[1:, :-1]) / self.cfg.dy
        ) * self.cfg.dx

        # 3. Sınır Koşulları
        if self.boundary_condition:
            self.boundary_condition.apply_boundary(self.ez, ez_prev_boundaries)

        # 4. Uyarım Kaynakları
        for source in self.sources:
            if isinstance(source, PointGaussianPulse2D):
                self.ez[source.pos_x, source.pos_y] += source.evaluate(current_step, self.cfg.dt)

        # 5. Anlık Zaman Adımı Kaydı
        self.ez_history[current_step, :, :] = self.ez.copy()

    def run(self) -> None:
        for n in range(self.cfg.time_steps):
            self.step(n)


# =============================================================================
# 4. GÖRSELLEŞTİRİCİ (HTML5 ANİMASYON SERVİSİ)
# =============================================================================
class FDTD2DVisualizer:
    @staticmethod
    def render_html5_animation(
        engine: FDTD2DEngineTMz,
        fps: int = 30,
        dpi: int = 100,
        embed_limit_mb: float = 100.0
    ) -> HTML:
        plt.rcParams['animation.embed_limit'] = embed_limit_mb

        cfg = engine.cfg
        fig, ax = plt.subplots(figsize=(7, 6), dpi=dpi)

        if not np.all(engine.eps_r == 1.0):
            ax.imshow(
                engine.eps_r.T,
                cmap='binary',
                origin='lower',
                alpha=0.3,
                extent=[0, cfg.nx, 0, cfg.ny]
            )

        vmax = np.max(np.abs(engine.ez_history)) * 0.8 or 1.0
        im = ax.imshow(
            engine.ez_history[0].T,
            cmap='seismic',
            origin='lower',
            vmin=-vmax,
            vmax=vmax,
            extent=[0, cfg.nx, 0, cfg.ny]
        )
        fig.colorbar(im, ax=ax, label=r'$E_z$ Genliği (V/m)')

        ax.set_xlabel("Uzay Ekseni X (Hücre)", fontsize=11)
        ax.set_ylabel("Uzay Ekseni Y (Hücre)", fontsize=11)

        def animate(frame: int):
            im.set_array(engine.ez_history[frame].T)
            ax.set_title(f"2D FDTD EM Dalga Yayılımı — Adım: {frame} / {cfg.time_steps}")
            return [im]

        anim = FuncAnimation(
            fig,
            animate,
            frames=cfg.time_steps,
            interval=1000 // fps,
            blit=True
        )
        plt.close(fig)

        return HTML(anim.to_jshtml())


# =============================================================================
# 5. ÇALIŞTIRMA VE DENEY BLOĞU
# =============================================================================
if __name__ == "__main__":
    # 1. 2D Izgara (150x150 Hücre, 300 Zaman Adımı)
    grid_cfg2d = GridConfig2D(nx=150, ny=150, time_steps=300)

    # 2. Silindirik Dielektrik Engel Ekleme (Eps_r = 3.0)
    eps_profile = np.ones((grid_cfg2d.nx, grid_cfg2d.ny), dtype=np.float64)
    cx, cy, radius = 100, 75, 20
    y_indices, x_indices = np.ogrid[:grid_cfg2d.ny, :grid_cfg2d.nx]
    mask = (x_indices - cx)**2 + (y_indices - cy)**2 <= radius**2
    eps_profile[mask.T] = 3.0

    # 3. Noktasal Uyarım Kaynağı ve Sınır Koşulu
    point_source = PointGaussianPulse2D(pos_x=40, pos_y=75)
    mur_abc2d = MurABC2D(grid_config=grid_cfg2d)

    # 4. Simülasyonu Çalıştırma
    engine2d = FDTD2DEngineTMz(
        grid_config=grid_cfg2d,
        sources=[point_source],
        boundary_condition=mur_abc2d,
        eps_r_profile=eps_profile
    )
    engine2d.run()

    # 5. Animasyonu Gösterme
    display(FDTD2DVisualizer.render_html5_animation(engine2d, fps=30))
```
